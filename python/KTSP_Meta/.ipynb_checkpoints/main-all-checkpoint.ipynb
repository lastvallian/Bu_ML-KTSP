```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append(r"C:\Users\Administrator\KTSP")
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from KTSP_Meta.pipe_line.KTSP_Meta_Pipeline import MetaPipeline
from models.multi_model_trainer import SKLearnModelWrapper
from models.multi_model_trainer import ShrinkageCentroidClassifier
from models.Ktsp_model import KTSPClassifier

models = {
    "Linear SVM": SKLearnModelWrapper(SVC(kernel="linear", C=1, probability=True)),
    "RBF SVM": SKLearnModelWrapper(SVC(kernel="rbf", C=5, gamma=0.05, probability=True)),
    "Decision Tree": SKLearnModelWrapper(DecisionTreeClassifier()),
    "Naive Bayes": SKLearnModelWrapper(GaussianNB()),
    "kNN": SKLearnModelWrapper(KNeighborsClassifier(n_neighbors=5)),
    "KTSP": KTSPClassifier(top_genes=300, top_pairs=10)
    #"Custom SVMTrainer":SKLearnModelWrapper(ShrinkageCentroidClassifier(shrinkage=2.0))
}


datasets = {
    #"Leukemia": r"D:\fitchburg\Bu_ML\KTSP\Data\Leukemia.txt"
    #"Colon": r"D:\fitchburg\Bu_ML\KTSP\Data\Colon.txt"
    #"DLBCL": r"D:\fitchburg\Bu_ML\CData\DLBCL.txt"
    #"LUNG": r"D:\fitchburg\Bu_ML\KTSP\Data\Lung.txt"
    #"GCM": r"D:\fitchburg\Bu_ML\KTSP\Data\GCM.txt"
    #"prostate1":r"D:\fitchburg\Bu_ML\KTSP\Data\prostate1.txt"
    #"prostate2":r"D:\fitchburg\Bu_ML\KTSP\Data\prostate2.txt"
    "prostate3": r"D:\fitchburg\Bu_ML\KTSP\Data\prostate3.txt"
}

pipeline = MetaPipeline(models)

all_results={}

for ds_name ,ds_path in datasets.items():
    print(f"\n=== Processing dataset")
    
    results=pipeline.run(ds_path)
    
    all_results[ds_name]=results
    
    for model_name,res in results.items():

        print(f"{model_name}: Accuracy = {res['accuracy']:.4f}")
```
