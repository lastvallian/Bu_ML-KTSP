```python
import numpy as np
import json
import os

def generate_test_csv():
    # 1. 自动获取项目根目录 (KTSP_Meta 层级)
    # __file__ 是当前脚本 utils/generate_test_csv.py 的路径
    # dirname 是 utils，再 dirname 就是 KTSP_Meta
    current_dir = os.getcwd() 
    
    # 如果当前就在 utils 里，向上跳一级到项目根目录
    if current_dir.endswith('utils'):
        project_root = os.path.dirname(current_dir)
    else:
        project_root = current_dir
    
    print(f"--- 自动定位项目根目录: {project_root} ---")

    # 2. 获取用户输入
    raw_input = input("请输入相对路径 (例如: saved_models/kNN/inference_config.json): ").strip().strip('"')

    # 3. 智能拼接：尝试直接找，或者在根目录下找
    # 这样无论你输入 ../saved_models 还是 saved_models 都能识别
    if os.path.isabs(raw_input):
        config_path = raw_input
    else:
        # 尝试相对于项目根目录拼接
        config_path = os.path.join(project_root, raw_input)

    # 4. 最终检查
    if not os.path.exists(config_path):
        print(f"❌ 依然找不到文件！")
        print(f"尝试搜索的绝对路径为: {config_path}")
        return False

    try:
        print(f"成功读取配置文件: {config_path}")
        with open(config_path, "r", encoding='utf-8') as f:
            config = json.load(f)
            feature_names = config.get("features", [])

        if not feature_names:
            print("❌ 错误：JSON 中没有特征数据")
            return False

        # 5. 生成测试数据
        data = np.random.rand(5, len(feature_names))
        df = pd.DataFrame(data, columns=feature_names)
        
        # 建议保存到项目根目录，方便前端上传
        output_path = os.path.join(project_root, "test_input.csv")
        df.to_csv(output_path, index=False)

        print("-" * 40)
        print(f"✅ 生成成功！")
        print(f"测试文件已保存至: {output_path}")
        print("-" * 40)
        return True

    except Exception as e:
        print(f"💥 运行异常: {e}")
        return False
```
